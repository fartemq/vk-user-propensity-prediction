```python
import pandas as pd

train = pd.read_parquet('train_clean (1).parquet')
```

```python
train.info()
```

```python
train.head()
```

```python
y = train['target']
X = train.drop(columns = ['index','target'])
print(X.shape)
print(y.shape)
```

##Рассуждения по поводу выбора модели

Посмотрев на данные видно, что целевая переменная - таргет, это столбец из 0 и 1, то есть соверишл или не совершил ползователь целевое действие, то есть перед нами стоит задача бинарной классификации, для ее решения логично воспользовать логистической регрессией, чтобы на выходе получать вероятность совершения пользователем целевого действия (склонность к целевому действию). Также в данном кейсе логично использовать L1-Регуляризацию, тк она поможет разрядить признаки, те занулить ненужыне.

## Шаг 1. Импорт библиотек и разделение данных

Перед обучением модели необходимо сделать две вещи:
1. Импортировать нужные инструменты для масштабирования, самой модели и оценки её качества.
2. Разбить данные на обучающую (Train) и валидационную (Validation) выборки. Это нужно, чтобы проверять качество модели на «незнакомых» ей данных и избежать переобучения. При этом используем стратификацию, чтобы сохранить баланс классов.


```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score
```

```python
X_train,X_val,y_train, y_val = train_test_split(X , y , test_size = 0.2 , stratify = y, random_state = 42)

print("Обучающая выборка:", X_train.shape)
print("Валидационная выборка:", X_val.shape)
```

## Шаг 2. Масштабирование признаков (Стандартизация)

Используем класс `StandardScaler` для приведения всех признаков к единому масштабу. 
Чтобы избежать утечки данных (Data Leakage), мы:
1. Обучаем скалер и трансформируем данные только на обучающей выборке (`X_train`).
2. Валидационную выборку (`X_val`) только трансформируем, используя среднее и дисперсию, посчитанные на обучающем наборе.


```python
scaler = StandardScaler()
X_train_scaler = scaler.fit_transform(X_train)
X_val_scaler = scaler.transform(X_val)
```

## Шаг 3. Создание и обучение линейной модели

Мы инициализируем логистическую регрессию с L1-регуляризацией (`penalty='l1'`). 
Это позволит алгоритму автоматически обнулить веса у шумных и избыточных зашифрованных признаков. 
Для работы с L1-регуляризацией на больших данных мы используем оптимизатор `solver='saga'`.


```python
model_1 = LogisticRegression( 
    penalty = 'l1' , 
    solver = 'saga', 
    max_iter = 20, 
    random_state = 42, 
    n_jobs = -1)

model_1.fit(X_train_scaler, y_train)
```

```python
preds = model_1.predict_proba(X_val_scaler)[ : , 1]
print("ROC-AUC:", roc_auc_score(y_val, preds))
```

## Увеличение итераций до 100

Для устранения предупреждения `ConvergenceWarning` и повышения качества предсказания, увеличим количество итераций `max_iter` до 100.


```python
model_100 = LogisticRegression( 
    penalty = 'l1' , 
    solver = 'saga', 
    max_iter = 100, 
    random_state = 42, 
    n_jobs = -1)

model_100.fit(X_train_scaler, y_train)
```

```python
preds_100 = model_100.predict_proba(X_val_scaler)[ : , 1]
print( "ROC - AUC :" , roc_auc_score(y_val,preds_100))
```

## Шаг 4.Нелинейный модели

Посторив линейную модель видно, что ей тяжело справится с задачей тк признаки мультиколерированны и их сликшом много, и отсуствуют прямые линейные зависимости.
Потому перейдем к следующему классу моделей - решающим деревьям, сперва попробуем решить задачу простым деревом, и будем использовать его как baseline после этого попробуем амсамбли и градиентный бустинг



```python
from sklearn.tree import DecisionTreeClassifier
model_tree = DecisionTreeClassifier( 
    max_depth= 10, 
    class_weight='balanced',  
    random_state=42)
model_tree.fit(X_train, y_train)
```

```python
from sklearn.metrics import roc_auc_score
preds_tree = model_tree.predict_proba(X_val)[:, 1]
roc_auc_tree = roc_auc_score(y_val, preds_tree)
print("-" * 50)
print(f"ROC-AUC одиночного дерева решений: {roc_auc_tree:.6f}")
print("-" * 50)
```

## Random forest

Одно дерево, очевидно, не дало сильного прироста метрики, поэтому перейдем к ансамблевым моделям и первой попробуем random forest



```python
from sklearn.ensemble import RandomForestClassifier
model_rdm_forest = RandomForestClassifier(
    n_estimators=300,
    max_depth=12,              # Немного уменьшили глубину, чтобы случайный лес не переобучался
    max_features=0.2,
    class_weight='balanced',   # Изменили под ROC-AUC для более стабильного веса классов
    n_jobs=-1,
    random_state=42
)

model_rdm_forest.fit(X_train, y_train)
```

```python
from sklearn.metrics import roc_auc_score

y_val_proba_tuned = model_rdm_forest.predict_proba(X_val)[:, 1]

roc_auc_tuned = roc_auc_score(y_val, y_val_proba_tuned)

print("-" * 50)
print(f"Новый ROC-AUC случайного леса с max_features=0.2: {roc_auc_tuned:.4f}")
print("-" * 50)
```

## Градиентный бустинг над решающими деревьями (GBDT)
Теперь попробуем самый мощный метод работы с табличными данными - градиентный бустинг. Для начала воспользуемся библиотекой Lightgbm,а затем CatBoost, далее объеденим их результаты и посмотрим повлияет ли это на метрику


```python
import lightgbm as lgb
from lightgbm import LGBMClassifier


lgb_simple = LGBMClassifier(
    objective="binary",
    metric="auc",          # Оптимизируем строго ROC-AUC
    boosting_type="gbdt",
    random_state=42,
    n_jobs=-1,
    learning_rate=0.05,    # Чуть повысили скорость для быстрого теста
    n_estimators=1000,     # Запас деревьев
    max_depth=6,           # Защита от переобучения на шуме
    num_leaves=31,
    feature_fraction=0.2   # Берем 20% случайных фичей из 1366, чтобы не оверфитить
)


lgb_simple.fit(X_train_scaler,y_train,eval_set=[(X_val_scaler, y_val)],
               callbacks=[lgb.early_stopping(stopping_rounds=50, verbose=False)]), # Остановит, если скор перестанет расти)


preds_proba_simple = lgb_simple.predict_proba(X_val_scaler)[:, 1]
roc_auc_simple = roc_auc_score(y_val, preds_proba_simple)

print("\n" + "=" * 50)
print(f" ROC-AUC НА ВСЕХ ПРИЗНАКАХ: {roc_auc_simple:.6f}")
print("=" * 50)
```

```python
from catboost import CatBoostClassifier

# Инициализируем CatBoost
cb_model = CatBoostClassifier(
    iterations=1500,
    learning_rate=0.03,
    depth=6,
    eval_metric='AUC',    
    random_seed=42,
    scale_pos_weight=10,  
    verbose=100,
    rsm=0.2, 
    thread_count=-1,
    l2_leaf_reg=5
)

# Обучаем
cb_model.fit(
    X_train_scaler, y_train,
    eval_set=(X_val_scaler, y_val),
    early_stopping_rounds=50
)

# Получаем его вероятности
preds_cb = cb_model.predict_proba(X_val_scaler)[:, 1]

```

```python
final_blend_preds = 0.5 * preds_proba_simple + 0.5 * preds_cb

blend_score = roc_auc_score(y_val, final_blend_preds)

print("=" * 50)
print(f"ФИНАЛЬНЫЙ ROC-AUC АНСАМБЛЯ (LGBM + CatBoost): {blend_score:.6f}")
print("=" * 50)
```

```python
import numpy as np
from sklearn.model_selection import StratifiedKFold



X_full = np.vstack([X_train, X_val])
y_full = np.concatenate([y_train, y_val])
n_splits = 5
skf = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=42)
oof_preds = np.zeros(len(X_full))
trained_models = []

for fold, (train_idx, val_idx) in enumerate(skf.split(X_full, y_full)):
    print(f"--- ОБУЧЕНИЕ НА ФОЛДЕ {fold + 1} ---")
    X_tr, y_tr = X_full[train_idx], y_full[train_idx]
    X_va, y_va = X_full[val_idx], y_full[val_idx]
    model = LGBMClassifier(
        objective="binary",
        boosting_type="gbdt",
        metric="auc",               
        random_state=42 + fold,     
        n_jobs=-1,
        learning_rate=0.05,         
        n_estimators=1000,          
        max_depth=6,                
        num_leaves=31,
        feature_fraction=0.2,             
    )
    model.fit(
        X_tr, y_tr,
        eval_set=[(X_va, y_va)],
        callbacks=[
            lgb.early_stopping(stopping_rounds=100, verbose=False), 
            lgb.log_evaluation(100)                                 
        ]
    )

    oof_preds[val_idx] = model.predict_proba(X_va)[:, 1]
    trained_models.append(model)
    print(f"Фолд {fold + 1} успешно завершен.\n")
    
final_oof_score = roc_auc_score(y_full, oof_preds)
print("=" * 60)
print(f"ЧЕСТНЫЙ СТРАТИФИЦИРОВАННЫЙ OOF ROC-AUC (5 фолдов): {final_oof_score:.6f}")
print("=" * 60)
```

## Сохранение


```python
import pandas as pd
import numpy as np

test_raw = pd.read_parquet('test_perfect.parquet')
test_indices = test_raw['index']

train_features = [col for col in train.columns if col not in ['index', 'target', 'y']]
X_test = test_raw[train_features]

num_cols = X_test.select_dtypes(include=[np.number]).columns
for col in num_cols:
    if X_test[col].isna().sum() > 0:
        X_test[col] = X_test[col].fillna(X_test[col].median() if not np.isnan(X_test[col].median()) else 0)

X_test_scaler = scaler.transform(X_test)

test_preds = np.zeros(len(X_test_scaler))
for model in trained_models:
    test_preds += model.predict_proba(X_test_scaler)[:, 1] / len(trained_models)

submission = pd.DataFrame({
    'index': test_indices,
    'target': test_preds
})

submission.to_csv('submission_lgbm_ensemble_1.csv', index=False)
print("Готово")
```
